```python
                                                                                 # ---------- elimar de csv----------
def eliminar_de_csv(archivo, indice_columna, valor_buscado):
    filas_nuevas = []
    try:
        with open(archivo, newline="", encoding="utf-8") as f:
            reader = csv.reader(f)
            for fila in reader:
                if fila[indice_columna] != valor_buscado:
                    filas_nuevas.append(fila)
        with open(archivo, "w", newline="", encoding="utf-8") as f:
            writer = csv.writer(f)
            writer.writerows(filas_nuevas)
    except FileNotFoundError:
        pass
```
